```python
# 1. Create a separate copy to preserve the original dataset
orders_clean = orders.copy()

# 2. Identify and remove duplicate order records
duplicate_order_rows = orders_clean.duplicated(
    subset="order_id", keep=False
)

print("Records with repeated order IDs:")
display(orders_clean.loc[duplicate_order_rows].sort_values("order_id"))

rows_before = len(orders_clean)

# Keep the first occurrence of each order ID
orders_clean = (
    orders_clean
    .drop_duplicates(subset="order_id", keep="first")
    .reset_index(drop=True)
)

duplicates_removed = rows_before - len(orders_clean)

# Standardize spaces and capitalization
orders_clean["order_status"] = (
    orders_clean["order_status"]
    .astype("string")
    .str.strip()
    .str.title()
)

print("Standardized order statuses:")
display(orders_clean["order_status"].value_counts(dropna=False))

# 3. Convert order amounts and date columns
amount_before = orders_clean["order_amount"].copy()

orders_clean["order_amount"] = pd.to_numeric(
    amount_before, errors="coerce"
)

invalid_amounts = (
    amount_before.notna() & orders_clean["order_amount"].isna()
).sum()

date_columns = [
    "order_date",
    "promised_delivery_date",
    "actual_delivery_date"
]

invalid_date_counts = {}

for column in date_columns:
    original_dates = orders_clean[column].astype("string").str.strip()

    # Treat empty strings as missing
    original_dates = original_dates.replace("", pd.NA)

    orders_clean[column] = pd.to_datetime(
        original_dates,
        format="%Y-%m-%d",
        errors="coerce"
    )

    # Count supplied values that could not be converted
    invalid_date_counts[column] = (
        original_dates.notna() & orders_clean[column].isna()
    ).sum()

# 4. Investigate missing delivery dates by order status
delivery_missing_summary = (
    orders_clean
    .assign(
        missing_delivery_date=
        orders_clean["actual_delivery_date"].isna()
    )
    .groupby("order_status", dropna=False)
    .agg(
        total_orders=("order_id", "size"),
        missing_delivery_dates=("missing_delivery_date", "sum")
    )
)

print("\nMissing delivery dates by status:")
display(delivery_missing_summary)

# Assumption: these statuses do not require a completed delivery date
valid_missing_statuses = ["Pending", "Shipped", "Cancelled"]

missing_delivery = orders_clean["actual_delivery_date"].isna()

valid_missing_delivery = (
    missing_delivery
    & orders_clean["order_status"].isin(valid_missing_statuses)
)

delivery_date_issues = orders_clean.loc[
    missing_delivery & ~valid_missing_delivery
].copy()

print("Missing delivery dates requiring investigation:")
display(delivery_date_issues)

# Summarize cleaning results
cleaning_summary = pd.Series({
    "original_records": len(orders),
    "duplicate_order_records_removed": duplicates_removed,
    "cleaned_records": len(orders_clean),
    "unique_orders_after_cleaning": orders_clean["order_id"].nunique(),
    "invalid_order_amounts": invalid_amounts,
    "missing_order_amounts": orders_clean["order_amount"].isna().sum(),
    "non_positive_order_amounts": orders_clean["order_amount"].le(0).sum(),
    "valid_missing_delivery_dates": valid_missing_delivery.sum(),
    "missing_delivery_dates_to_investigate": len(delivery_date_issues)
}, name="count")

print("\nCleaning summary:")
display(cleaning_summary)

print("Invalid date values converted to NaT:")
display(pd.Series(invalid_date_counts, name="invalid_date_count"))

print("Cleaned data types:")
display(orders_clean.dtypes)

display(orders_clean.head())
```

```python
# 1. Create separate copies
refunds_clean = refunds.copy()
complaints_clean = complaints.copy()

# 2. Inspect duplicate identifiers before removing them
print("Duplicate refund IDs:")
display(
    refunds_clean.loc[
        refunds_clean.duplicated("refund_id", keep=False)
    ].sort_values("refund_id")
)

print("Duplicate complaint IDs:")
display(
    complaints_clean.loc[
        complaints_clean.duplicated("complaint_id", keep=False)
    ].sort_values("complaint_id")
)

# Check for conflicting records sharing the same identifier
refund_id_conflicts = (
    refunds_clean.drop_duplicates()
    .duplicated("refund_id", keep=False)
)

complaint_id_conflicts = (
    complaints_clean.drop_duplicates()
    .duplicated("complaint_id", keep=False)
)

# Stop for investigation if duplicates contain conflicting information
if refund_id_conflicts.any() or complaint_id_conflicts.any():
    raise ValueError(
        "Conflicting records share an identifier. "
        "Review them before choosing which record to keep."
    )

refunds_before = len(refunds_clean)
complaints_before = len(complaints_clean)

refunds_clean = (
    refunds_clean.drop_duplicates("refund_id")
    .reset_index(drop=True)
)

complaints_clean = (
    complaints_clean.drop_duplicates("complaint_id")
    .reset_index(drop=True)
)

# 3. Standardize identifiers, statuses, and complaint categories
for column in ["refund_id", "order_id"]:
    refunds_clean[column] = (
        refunds_clean[column].astype("string").str.strip()
        .replace("", pd.NA)
    )

for column in ["complaint_id", "order_id", "customer_id"]:
    complaints_clean[column] = (
        complaints_clean[column].astype("string").str.strip()
        .replace("", pd.NA)
    )

refunds_clean["refund_status"] = (
    refunds_clean["refund_status"]
    .astype("string").str.strip().str.title()
    .replace("", pd.NA)
)

for column in ["complaint_status", "complaint_type"]:
    complaints_clean[column] = (
        complaints_clean[column]
        .astype("string").str.strip().str.title()
        .replace("", pd.NA)
    )

# Convert monetary amounts and track invalid supplied values
original_amounts = refunds_clean["refund_amount"].copy()

refunds_clean["refund_amount"] = pd.to_numeric(
    original_amounts, errors="coerce"
)

invalid_refund_amounts = (
    original_amounts.notna()
    & refunds_clean["refund_amount"].isna()
).sum()

# Convert dates and track invalid supplied values
invalid_date_counts_m4 = {}

for table, column in [
    (refunds_clean, "refund_date"),
    (complaints_clean, "complaint_date")
]:
    original_dates = (
        table[column].astype("string").str.strip()
        .replace("", pd.NA)
    )

    table[column] = pd.to_datetime(
        original_dates,
        format="%Y-%m-%d",
        errors="coerce"
    )

    invalid_date_counts_m4[column] = (
        original_dates.notna() & table[column].isna()
    ).sum()

# 4. Identify missing and unmatched order references
known_order_ids = (
    orders_clean["order_id"].astype("string").str.strip()
    .dropna()
)

refunds_missing_order_id = refunds_clean.loc[
    refunds_clean["order_id"].isna()
].copy()

complaints_missing_order_id = complaints_clean.loc[
    complaints_clean["order_id"].isna()
].copy()

refunds_unmatched_orders = refunds_clean.loc[
    refunds_clean["order_id"].notna()
    & ~refunds_clean["order_id"].isin(known_order_ids)
].copy()

complaints_unmatched_orders = complaints_clean.loc[
    complaints_clean["order_id"].notna()
    & ~complaints_clean["order_id"].isin(known_order_ids)
].copy()

quality_summary_m4 = pd.Series({
    "duplicate_refunds_removed": refunds_before - len(refunds_clean),
    "duplicate_complaints_removed": (
        complaints_before - len(complaints_clean)
    ),
    "missing_refund_ids": refunds_clean["refund_id"].isna().sum(),
    "missing_complaint_ids": complaints_clean["complaint_id"].isna().sum(),
    "refunds_missing_order_id": len(refunds_missing_order_id),
    "complaints_missing_order_id": len(complaints_missing_order_id),
    "refunds_with_unmatched_orders": len(refunds_unmatched_orders),
    "complaints_with_unmatched_orders": len(complaints_unmatched_orders),
    "invalid_refund_amounts": invalid_refund_amounts,
    "missing_refund_amounts": refunds_clean["refund_amount"].isna().sum(),
    "non_positive_refund_amounts": (
        refunds_clean["refund_amount"].le(0).sum()
    ),
    "invalid_refund_dates": invalid_date_counts_m4["refund_date"],
    "invalid_complaint_dates": invalid_date_counts_m4["complaint_date"]
}, name="count")

print("\nData quality summary:")
display(quality_summary_m4)

print("Missing values in cleaned refunds:")
display(refunds_clean.isna().sum())

print("Missing values in cleaned complaints:")
display(complaints_clean.isna().sum())

print("Refunds with unmatched order references:")
display(refunds_unmatched_orders)

print("Complaints with unmatched order references:")
display(complaints_unmatched_orders)

print("Standardized refund statuses:")
display(refunds_clean["refund_status"].value_counts(dropna=False))

print("Standardized complaint statuses:")
display(complaints_clean["complaint_status"].value_counts(dropna=False))

print("Cleaned refund dataset:")
display(refunds_clean)

print("Cleaned complaint dataset:")
display(complaints_clean)
```

```python
# 1. Order status distribution
order_status_summary = (
    orders_clean["order_status"]
    .value_counts(dropna=False)
    .rename_axis("order_status")
    .reset_index(name="order_count")
)

order_status_summary["percentage"] = (
    order_status_summary["order_count"] / len(orders_clean) * 100
).round(2)

print("Order status distribution:")
display(order_status_summary)

# Total order amount by category, across all order statuses
category_order_amounts = (
    orders_clean
    .groupby("product_category", dropna=False)["order_amount"]
    .sum(min_count=1)
    .reset_index(name="total_order_amount")
    .sort_values("total_order_amount", ascending=False)
    .reset_index(drop=True)
)

print("Total order amount by product category (all statuses):")
display(category_order_amounts)

# 2. Select successfully delivered orders
delivered_orders = orders_clean.loc[
    orders_clean["order_status"].eq("Delivered")
].copy()

# Identify records missing either date needed for comparison
incomplete_dates = delivered_orders[
    ["promised_delivery_date", "actual_delivery_date"]
].isna().any(axis=1)

delivered_orders_incomplete = delivered_orders.loc[
    incomplete_dates
].copy()

# Analyze only delivered orders with both dates available
delivery_performance = delivered_orders.loc[
    ~incomplete_dates
].copy()

delivery_performance["delivery_delay_days"] = (
    delivery_performance["actual_delivery_date"]
    - delivery_performance["promised_delivery_date"]
).dt.days

delivery_performance["is_delayed"] = (
    delivery_performance["delivery_delay_days"] > 0
)

# Negative = early; zero = on time; positive = late
print("Delivery date comparisons:")
display(
    delivery_performance[
        [
            "order_id",
            "customer_city",
            "promised_delivery_date",
            "actual_delivery_date",
            "delivery_delay_days",
            "is_delayed"
        ]
    ].head()
)

# 3. Identify delayed deliveries
delayed_orders = delivery_performance.loc[
    delivery_performance["is_delayed"]
].copy()

# Count all assessable deliveries and delayed deliveries by city
city_delivery_summary = (
    delivery_performance
    .groupby("customer_city", dropna=False)
    .agg(
        assessed_deliveries=("order_id", "size"),
        delayed_deliveries=("is_delayed", "sum")
    )
)

city_delivery_summary["delayed_percentage"] = (
    city_delivery_summary["delayed_deliveries"]
    / city_delivery_summary["assessed_deliveries"] * 100
).round(2)

# Average and maximum delay among late deliveries only
city_delay_stats = (
    delayed_orders
    .groupby("customer_city", dropna=False)
    .agg(
        average_delay_days=("delivery_delay_days", "mean"),
        maximum_delay_days=("delivery_delay_days", "max")
    )
)

city_delivery_summary = (
    city_delivery_summary
    .join(city_delay_stats)
    .reset_index()
    .sort_values(
        ["delayed_deliveries", "customer_city"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

city_delivery_summary["average_delay_days"] = (
    city_delivery_summary["average_delay_days"].round(2)
)

print("Delivery performance by city:")
display(city_delivery_summary)

# 4. Five cities with the most delayed deliveries
top_5_delayed_cities = city_delivery_summary.loc[
    city_delivery_summary["delayed_deliveries"] > 0
].head(5).copy()

print("Five cities with the highest number of delayed deliveries:")
display(top_5_delayed_cities)

print("Delivered orders with incomplete delivery information:")
display(delivered_orders_incomplete)

print(f"Total delivered orders: {len(delivered_orders)}")
print(f"Delivered orders assessed: {len(delivery_performance)}")
print(f"Delayed deliveries: {len(delayed_orders)}")
print(
    "Delivered orders with incomplete dates:",
    len(delivered_orders_incomplete)
)
```

```python
# 1. Summarize refund requests at the order level
# Keep records with missing order IDs separate for investigation
refunds_with_order_id = refunds_clean.loc[
    refunds_clean["order_id"].notna()
].copy()

refund_summary = (
    refunds_with_order_id
    .groupby("order_id")
    .agg(
        refund_request_count=("refund_id", "size"),
        total_requested_refund_amount=(
            "refund_amount",
            lambda amounts: amounts.sum(min_count=1)
        )
    )
)

# 2. Identify unresolved requests
unresolved_refunds = refunds_clean.loc[
    refunds_clean["refund_status"].isin(["Pending", "Failed"])
].copy()

unresolved_refund_counts = (
    unresolved_refunds
    .dropna(subset=["order_id"])
    .groupby("order_id")
    .size()
    .rename("unresolved_refund_count")
)

refund_summary = (
    refund_summary
    .join(unresolved_refund_counts)
    .reset_index()
)

refund_summary["unresolved_refund_count"] = (
    refund_summary["unresolved_refund_count"]
    .fillna(0)
    .astype(int)
)

print("Order-level refund summary:")
display(refund_summary.head())

print("Pending and failed refund requests:")
display(unresolved_refunds)

# 3. Left join to retain every cleaned order
orders_refunds = orders_clean.merge(
    refund_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Identify orders without refund requests before filling counts
no_refund_requests = orders_refunds["refund_request_count"].isna()

for column in ["refund_request_count", "unresolved_refund_count"]:
    orders_refunds[column] = (
        orders_refunds[column].fillna(0).astype(int)
    )

# Zero means no refund was requested.
# Missing amounts on existing requests remain missing.
orders_refunds.loc[
    no_refund_requests, "total_requested_refund_amount"
] = 0

# 4. Analyze refund amounts by status
# This includes all cleaned refund records, even unmatched references
refund_amounts_by_status = (
    refunds_clean
    .groupby("refund_status", dropna=False)
    .agg(
        refund_request_count=("refund_id", "size"),
        known_amount_count=("refund_amount", "count"),
        total_requested_amount=(
            "refund_amount",
            lambda amounts: amounts.sum(min_count=1)
        ),
        average_requested_amount=("refund_amount", "mean")
    )
    .reset_index()
)

refund_amounts_by_status["missing_amount_count"] = (
    refund_amounts_by_status["refund_request_count"]
    - refund_amounts_by_status["known_amount_count"]
)

refund_amounts_by_status["average_requested_amount"] = (
    refund_amounts_by_status["average_requested_amount"].round(2)
)

print("Refund amounts by status:")
display(refund_amounts_by_status)

# Rank matched orders with at least one refund request
top_5_refund_orders = (
    orders_refunds.loc[
        orders_refunds["refund_request_count"].gt(0)
        & orders_refunds["total_requested_refund_amount"].notna()
    ]
    .sort_values(
        ["total_requested_refund_amount", "order_id"],
        ascending=[False, True]
    )
    .head(5)
    .copy()
)

print("Five matched orders with the highest requested refund amounts:")
display(
    top_5_refund_orders[
        [
            "order_id",
            "customer_id",
            "order_amount",
            "refund_request_count",
            "total_requested_refund_amount",
            "unresolved_refund_count"
        ]
    ]
)

print("Combined order-refund dataset:")
display(orders_refunds)

# Confirm that the merge retained exactly one row per order
assert len(orders_refunds) == len(orders_clean)
assert orders_refunds["order_id"].is_unique
```

```python
# 1. Analyze complaint types and resolution statuses
complaints_by_type_status = (
    complaints_clean
    .groupby(["complaint_type", "complaint_status"], dropna=False)
    .size()
    .reset_index(name="complaint_count")
    .sort_values("complaint_count", ascending=False)
)

print("Complaints by type and resolution status:")
display(complaints_by_type_status)

# 2. Count complaints per order
# Missing order references cannot be assigned to an order
complaint_summary = (
    complaints_clean
    .dropna(subset=["order_id"])
    .groupby("order_id")
    .size()
    .reset_index(name="complaint_count")
)

repeated_complaint_orders = (
    complaint_summary.loc[
        complaint_summary["complaint_count"] > 1
    ]
    .sort_values(
        ["complaint_count", "order_id"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("Complaint count per order:")
display(complaint_summary)

print("Orders with repeated complaints:")
display(repeated_complaint_orders)

# 3. Rank customers by complaint frequency
customer_complaint_ranking = (
    complaints_clean
    .dropna(subset=["customer_id"])
    .groupby("customer_id")
    .size()
    .reset_index(name="complaint_count")
    .sort_values(
        ["complaint_count", "customer_id"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

customer_complaint_ranking["rank"] = (
    customer_complaint_ranking["complaint_count"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

customers_multiple_complaints = (
    customer_complaint_ranking.loc[
        customer_complaint_ranking["complaint_count"] > 1
    ].copy()
)

print("Customers with multiple complaints, ranked by frequency:")
display(customers_multiple_complaints)

# 4. Find the five most common known complaint categories
complaint_category_counts = (
    complaints_clean
    .groupby("complaint_type", dropna=False)
    .size()
    .reset_index(name="complaint_count")
    .sort_values(
        ["complaint_count", "complaint_type"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

top_5_complaint_categories = (
    complaint_category_counts
    .dropna(subset=["complaint_type"])
    .head(5)
    .copy()
)

print("Five most common complaint categories:")
display(top_5_complaint_categories)

# Identify unresolved complaints
unresolved_complaints = complaints_clean.loc[
    complaints_clean["complaint_status"].isin(["Open", "In Progress"])
].copy()

unresolved_complaint_summary = (
    unresolved_complaints
    .groupby(["complaint_type", "complaint_status"], dropna=False)
    .size()
    .reset_index(name="complaint_count")
    .sort_values("complaint_count", ascending=False)
)

# Keep an order-level unresolved count for later milestones
unresolved_complaint_counts = (
    unresolved_complaints
    .dropna(subset=["order_id"])
    .groupby("order_id")
    .size()
    .reset_index(name="unresolved_complaint_count")
)

complaint_summary = complaint_summary.merge(
    unresolved_complaint_counts,
    on="order_id",
    how="left",
    validate="one_to_one"
)

complaint_summary["unresolved_complaint_count"] = (
    complaint_summary["unresolved_complaint_count"]
    .fillna(0)
    .astype(int)
)

print("Unresolved complaints by category and status:")
display(unresolved_complaint_summary)

print("Unresolved complaint records:")
display(unresolved_complaints)

print(f"Total cleaned complaints: {len(complaints_clean)}")
print(f"Orders with repeated complaints: {len(repeated_complaint_orders)}")
print(f"Customers with multiple complaints: {len(customers_multiple_complaints)}")
print(f"Total unresolved complaints: {len(unresolved_complaints)}")

# Flag statuses outside the expected set for separate review
unexpected_complaint_statuses = complaints_clean.loc[
    ~complaints_clean["complaint_status"]
    .isin(["Open", "In Progress", "Resolved"])
].copy()

print("Complaints with missing or unexpected statuses:")
display(unexpected_complaint_statuses)
```

```python
# 1. Prepare order-level issue summaries
refund_order_summary = refund_summary.copy()

complaint_order_summary = complaint_summary.copy()

# Count complaints specifically marked Open
open_complaint_counts = (
    complaints_clean.loc[
        complaints_clean["complaint_status"].eq("Open")
    ]
    .dropna(subset=["order_id"])
    .groupby("order_id")
    .size()
    .reset_index(name="open_complaint_count")
)

complaint_order_summary = complaint_order_summary.merge(
    open_complaint_counts,
    on="order_id",
    how="left",
    validate="one_to_one"
)

complaint_order_summary["open_complaint_count"] = (
    complaint_order_summary["open_complaint_count"]
    .fillna(0)
    .astype(int)
)

# 2. Combine summaries with cleaned orders
order_issues = (
    orders_clean
    .merge(
        refund_order_summary,
        on="order_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        complaint_order_summary,
        on="order_id",
        how="left",
        validate="one_to_one"
    )
)

# 3. Fill missing counts for orders without matching issue records
no_refund_requests = order_issues["refund_request_count"].isna()

count_columns = [
    "refund_request_count",
    "unresolved_refund_count",
    "complaint_count",
    "unresolved_complaint_count",
    "open_complaint_count"
]

for column in count_columns:
    order_issues[column] = (
        order_issues[column].fillna(0).astype(int)
    )

# Set amount to zero only when no refund request exists.
# Preserve missing amounts on existing refund requests.
order_issues.loc[
    no_refund_requests, "total_requested_refund_amount"
] = 0

# 4. Validate one record per cleaned order and complete retention
assert orders_clean["order_id"].notna().all()
assert orders_clean["order_id"].is_unique

assert len(order_issues) == len(orders_clean)
assert order_issues["order_id"].is_unique
assert not order_issues.duplicated().any()

assert set(order_issues["order_id"]) == set(orders_clean["order_id"])

assert (
    order_issues["unresolved_refund_count"]
    <= order_issues["refund_request_count"]
).all()

assert (
    order_issues["open_complaint_count"]
    <= order_issues["unresolved_complaint_count"]
).all()

assert (
    order_issues["unresolved_complaint_count"]
    <= order_issues["complaint_count"]
).all()

print("Validation passed: one record per cleaned order.")
print("Total orders retained:", len(order_issues))
print(
    "Orders with refund requests:",
    order_issues["refund_request_count"].gt(0).sum()
)
print(
    "Orders with complaints:",
    order_issues["complaint_count"].gt(0).sum()
)

print("\nConsolidated order issue dataset:")
display(order_issues)
```

```python
# Work on a copy of the consolidated dataset
order_priority = order_issues.copy()

# 1. Identify the three issue conditions
is_delivered = order_priority["order_status"].eq("Delivered")

complete_delivery_dates = order_priority[
    ["promised_delivery_date", "actual_delivery_date"]
].notna().all(axis=1)

# Calculate delivery delay only for assessable delivered orders
order_priority["delivery_delay_days"] = (
    (
        order_priority["actual_delivery_date"]
        - order_priority["promised_delivery_date"]
    )
    .dt.days
    .where(is_delivered & complete_delivery_dates)
)

order_priority["delivered_over_3_days_late"] = (
    is_delivered
    & complete_delivery_dates
    & order_priority["delivery_delay_days"].gt(3)
).fillna(False)

order_priority["has_unresolved_refund"] = (
    order_priority["unresolved_refund_count"].gt(0)
)

order_priority["has_multiple_complaints"] = (
    order_priority["complaint_count"].gt(1)
)

# Flag delivered orders missing a promised or actual delivery date
order_priority["incomplete_delivery_information"] = (
    is_delivered & ~complete_delivery_dates
).fillna(False)

# 2. Count how many conditions each order satisfies
issue_columns = [
    "delivered_over_3_days_late",
    "has_unresolved_refund",
    "has_multiple_complaints"
]

order_priority["issue_condition_count"] = (
    order_priority[issue_columns].astype(int).sum(axis=1)
)

# 3. Assign priority
order_priority["priority"] = (
    order_priority["issue_condition_count"]
    .map({
        0: "Low",
        1: "Medium",
        2: "High",
        3: "High"
    })
)

# 4. Summarize priority distribution
priority_summary = (
    order_priority["priority"]
    .value_counts()
    .reindex(["High", "Medium", "Low"], fill_value=0)
    .rename_axis("priority")
    .reset_index(name="order_count")
)

priority_summary["percentage"] = (
    priority_summary["order_count"] / len(order_priority) * 100
).round(2)

print("Priority distribution:")
display(priority_summary)

# Rank High-priority orders by number of conditions,
# then unresolved refunds, complaints, and delivery delay
high_priority_orders = (
    order_priority.loc[order_priority["priority"].eq("High")]
    .sort_values(
        [
            "issue_condition_count",
            "unresolved_refund_count",
            "complaint_count",
            "delivery_delay_days",
            "order_id"
        ],
        ascending=[False, False, False, False, True],
        na_position="last"
    )
    .copy()
)

top_10_high_priority_orders = high_priority_orders.head(10).copy()

print("Top ten High-priority orders:")
display(
    top_10_high_priority_orders[
        [
            "order_id",
            "customer_id",
            "customer_city",
            "delivery_delay_days",
            "unresolved_refund_count",
            "complaint_count",
            *issue_columns,
            "issue_condition_count",
            "priority",
            "incomplete_delivery_information"
        ]
    ]
)

# Separate review queue for incomplete delivery information
delivery_information_review = order_priority.loc[
    order_priority["incomplete_delivery_information"]
].copy()

print("Delivered orders requiring delivery information review:")
display(delivery_information_review)

# Validate priority rules and order retention
assert order_priority["order_id"].is_unique
assert len(order_priority) == len(order_issues)
assert order_priority["issue_condition_count"].between(0, 3).all()
assert order_priority["priority"].notna().all()
assert priority_summary["order_count"].sum() == len(order_issues)
```

```python
# 1. Select essential details for the final report
report_columns = [
    "order_id",
    "customer_id",
    "customer_city",
    "product_category",
    "order_status",
    "order_date",
    "promised_delivery_date",
    "actual_delivery_date",
    "order_amount",
    "delivery_delay_days",
    "incomplete_delivery_information",
    "refund_request_count",
    "total_requested_refund_amount",
    "unresolved_refund_count",
    "complaint_count",
    "open_complaint_count",
    "unresolved_complaint_count",
    "delivered_over_3_days_late",
    "has_unresolved_refund",
    "has_multiple_complaints",
    "issue_condition_count",
    "priority"
]

final_report = order_priority[report_columns].copy()

# 2. Sort High, then Medium, then Low priority
final_report["priority"] = pd.Categorical(
    final_report["priority"],
    categories=["High", "Medium", "Low"],
    ordered=True
)

final_report = (
    final_report
    .sort_values(
        [
            "priority",
            "issue_condition_count",
            "unresolved_refund_count",
            "complaint_count",
            "delivery_delay_days",
            "order_id"
        ],
        ascending=[True, False, False, False, False, True],
        na_position="last"
    )
    .reset_index(drop=True)
)

# 3. Prepare the support-priority summary
support_priority_summary = (
    final_report
    .groupby("priority", observed=False)
    .agg(
        order_count=("order_id", "size"),
        total_order_amount=(
            "order_amount",
            lambda values: values.sum(min_count=1)
        ),
        refund_request_count=("refund_request_count", "sum"),
        total_requested_refund_amount=(
            "total_requested_refund_amount",
            lambda values: values.sum(min_count=1)
        ),
        unresolved_refund_count=("unresolved_refund_count", "sum"),
        complaint_count=("complaint_count", "sum"),
        unresolved_complaint_count=("unresolved_complaint_count", "sum"),
        orders_over_3_days_late=("delivered_over_3_days_late", "sum"),
        orders_with_incomplete_delivery_info=(
            "incomplete_delivery_information", "sum"
        )
    )
    .reset_index()
)

total_orders = len(final_report)

support_priority_summary["order_percentage"] = (
    support_priority_summary["order_count"]
    .div(total_orders if total_orders else float("nan"))
    .mul(100)
    .round(2)
)

# Validate report consistency before exporting
assert final_report["order_id"].is_unique
assert len(final_report) == len(orders_clean)
assert support_priority_summary["order_count"].sum() == total_orders

# Export files into the Colab working directory
final_report.to_csv(
    "order_support_report.csv",
    index=False,
    date_format="%Y-%m-%d"
)

support_priority_summary.to_excel(
    "support_priority_summary.xlsx",
    index=False
)

print("Final order report:")
display(final_report)

print("Support-priority summary:")
display(support_priority_summary)

print("Exported: order_support_report.csv")
print("Exported: support_priority_summary.xlsx")

# 4. Generate evidence-based findings and recommendations
high_count = final_report["priority"].eq("High").sum()
high_percentage = (
    high_count / total_orders * 100 if total_orders else 0
)

assessed_deliveries = final_report["delivery_delay_days"].notna().sum()
severely_late_count = final_report["delivered_over_3_days_late"].sum()

unresolved_refund_total = final_report["unresolved_refund_count"].sum()
orders_with_unresolved_refunds = (
    final_report["unresolved_refund_count"].gt(0).sum()
)

repeat_complaint_orders = final_report["has_multiple_complaints"].sum()
unresolved_complaint_total = (
    final_report["unresolved_complaint_count"].sum()
)

incomplete_delivery_count = (
    final_report["incomplete_delivery_information"].sum()
)

business_findings = [
    (
        f"1. {high_count} orders ({high_percentage:.1f}%) are High priority "
        "because they satisfy at least two issue conditions. "
        "Recommendation: review these cases first using the sorted report."
    ),
    (
        f"2. {severely_late_count} of {assessed_deliveries} assessable "
        f"delivered orders arrived more than three days late; "
        f"{incomplete_delivery_count} delivered orders have incomplete dates. "
        "Recommendation: investigate severe delays and verify missing dates "
        "before drawing conclusions about delivery performance."
    ),
    (
        f"3. {unresolved_refund_total} pending or failed refund requests "
        f"are linked to {orders_with_unresolved_refunds} known orders. "
        "Recommendation: ask Finance to verify pending requests and "
        "investigate failed requests before retrying payments."
    ),
    (
        f"4. {repeat_complaint_orders} known orders have multiple complaints, "
        f"with {unresolved_complaint_total} unresolved complaints across "
        "all known orders. Recommendation: assign case ownership to "
        "coordinate follow-up and review repeated complaint histories."
    )
]

print("\nBusiness findings and recommendations:")
for finding in business_findings:
    print(f"\n{finding}")

print(
    "\nReport scope: one row per cleaned order. Issue records with missing "
    "or unmatched order references are excluded from order-level totals "
    "and require separate review. Requested refund amounts include all "
    "statuses and do not represent money actually paid. Amount totals "
    "sum available values and may be incomplete if amounts are missing."
)
```
